```python
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt
```

```python

#data={'skill':['writing', 'coding'], '}

df=pd.DataFrame(columns=['skill', 'component', 'weight'])

df.loc[0]=['Writing', 'Idea', 50]
df.loc[1]=['Writing', 'Implementation', 50]
df.loc[2]=['Coding', 'Idea', 90]
df.loc[3]=['Coding', 'Implementation', 10]
df
```

```python
skills={'Writing':{'Idea':50, 'Implementation':50}, 
        'Coding':{'Idea':90, 'Implementation':10}}
component=['Idea', 'Implementation']

df=pd.DataFrame(columns=['skill', 'component'])

for s in skills:
    for i in range(skills[s]['Idea']):
        df.loc[len(df)]=[s, 'Idea']
    for i in range(skills[s]['Implementation']):
        df.loc[len(df)]=[s, 'Implementation']

df
```

```python
sns.histplot(data=df, y='skill', hue='component', multiple='fill');
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.despine(top=True, right=True)

sns.barplot(y=['Writing', 'Coding'], x=[50, 10], color='#0072b5ff', edgecolor='grey', ax=ax)

sns.barplot(y=['Writing', 'Coding'], x=[-50, -90], color='#e18727ff', edgecolor='grey', ax=ax)

ax.axvline(x=0, color='black', alpha=0.5)

#tick and labels
ax.set(xlabel='Value', xlim=(-100,100));
ax.set_xticks([t for t in ax.get_xticks()])
new_labels=[abs(t) for t in ax.get_xticks()]
ax.set_xticklabels(new_labels)
ax.tick_params(left=False, bottom=False)

#annotation
ax.text(-65, 0, 'Idea', fontdict={'color':'#e18727ff', 'weight':'bold', 'size':15});
ax.text(20, 1, 'Implementation', fontdict={'color':'#0072b5ff', 'weight':'bold', 'size':15});
```
